```python
''' 
 Theory Questions ::

 1. What is the difference between function and a method in python?
       Ans:-  
         The key difference 
         between a function and a method in Python lies in their association with objects or classes. Functions are independent blocks of code,
         while methods are functions associated with objects of a class.
         Functions are independent blocks of code that can be called from anywhere, while methods are tied to objects or classes and need an 
         object or class instance to be invoked.
         E.g.
            Function :  
            def add_numbers(a, b):
            """This function takes two numbers as input and returns their sum."""
             return a + b
            
            result = add_numbers(5, 3)
            print(result) # Output: 8

            Method : 
            class Dog:
            def __init__(self, name, breed):
            self.name = name
            self.breed = breed
            
            def bark(self): # Instance method
            print("Woof!")
            
            @classmethod
            def create_generic_dog(cls): # Class method
            return cls("Generic Dog", "Unknown")
            
            @staticmethod
            def is_animal(): # Static method
            print("True")
            
            # Creating an instance of the Dog class
            my_dog = Dog("Buddy", "Golden Retriever")
            
            # Calling the bark method on the my_dog object
            my_dog.bark() # Output: Woof!
            
            # Calling the class method
            generic_dog = Dog.create_generic_dog()
            print(generic_dog.name) # Output: Generic Dog
            
            # Calling the static method
            Dog.is_animal() # Output: True
 2. Explain the concept of function arguments and parameters in python.
     Ans:-
       Parameters
       are variables listed inside the parentheses in the function definition. They act as placeholders for the values that the function will receive. 
       Arguments
       are the actual values that are passed to the function when it is called. These values are assigned to the corresponding parameters within the
       function's scope.
       E.g.
        def add_numbers(num1, num2):
         sum = num1 + num2
        return sum
        
        result = add_numbers(5, 3)
        print(result) # Output: 8
 3. What are the different ways to define and call a function in python?
    Ans:-
        In Python, functions can be defined and called in several ways:
        a. Standard Function Definition
        This is the most common way to define a function, using the def keyword, followed by the function name, parentheses for arguments, and a colon.
        The function body is indented.

        E.g.
        
        def greet(name):
            return "Hello, " + name + "!"
        
        # Calling the function
        message = greet("World")
        
        b. Lambda Functions
        These are small, anonymous functions defined with the lambda keyword. They can take any number of arguments but can only have one expression. 
        
        E.g.
        add = lambda x, y: x + y
        
        # Calling the lambda function
        result = add(5, 3)
        
        c. Nested Functions
        A function can be defined inside another function. The inner function has access to the outer function's scope.
        E.g.
        def outer_function(x):
            def inner_function(y):
                return x + y
            return inner_function
        
        # Calling the outer function returns the inner function
        closure = outer_function(10)
        
        # Calling the inner function through the closure
        result = closure(5)
        print(result)
        
        d. Recursive Functions
        A function can call itself. This is useful for tasks that can be broken down into smaller, self-similar subtasks.
        E.g.
        def factorial(n):
            if n == 0:
                return 1
            else:
                return n * factorial(n - 1)
        
        # Calling the recursive function
        result = factorial(5)
        print(result)
        
        e. Functions with Default Arguments
        Functions can have default values for arguments, which are used if the caller doesn't provide them.
        E.g.
        def power(base, exponent=2):
            return base ** exponent
        
        # Calling with both arguments
        result1 = power(3, 3)
        print(result1) # Output: 27
        
        # Calling with only the base argument
        result2 = power(3)
        print(result2)
 4. What is the purpose of the return statement in python function? 
        Ans:- The return statement in a Python function serves two primary purposes: 
        Exiting the function:
        When a return statement is encountered, the function's execution is immediately terminated, and control is passed back to the caller. 
        Any code within the function after the return statement will not be executed. 
        Returning a value:
        The return statement can optionally return a value to the caller. This value can be of any data type, including single values, multiple 
        values (as a tuple or list), or complex data structures. If no value is specified, the function returns None by default. 
        The returned value can then be used by the caller for further processing, assignment to variables, or as input to other functions.
        E.g.
        def add(x, y):
            return x + y
        
        result = add(5, 3)
        print(result)  # Output: 8
        
        def greet(name):
            print("Hello, " + name + "!")
            return  # Returns None implicitly
        
        return_value = greet("Alice") # Output: Hello, Alice!
        print(return_value) # Output: None
 5. What are iterators in python and how do they differ from itrables?
        Ans:- 
               An iterator is an object that contains a countable number of values. An iterator is an object that can be iterated upon, meaning that you
               can traverse through all the values. Technically, in Python, an iterator is an object which implements the iterator protocol, 
               which consist of the methods __iter__() and __next__() .
        
               Iterables and iterators are fundamental concepts in Python that facilitate looping and data traversal. They differ in their functionality
               and how they are used.
        An iterable is an object that can be looped over, meaning its elements can be accessed one at a time. Common examples of iterables include lists,
        tuples, strings, and dictionaries. Iterables implement the __iter__() method, which returns an iterator object.
        An iterator is an object that facilitates the actual iteration process. It maintains a state and produces the next value in the sequence when
        requested. Iterators implement the __iter__() and __next__() methods. The __next__() method returns the next element in the iteration and raises
        a StopIteration exception when no more elements are available. 
        In essence, an iterable is a container of data, while an iterator is a tool to access that data sequentially. An iterator is created from an
        iterable using the iter() function. Every iterator is also an iterable, but not every iterable is an iterator. 
        
        E.g.
        my_list = [1, 2, 3]  # my_list is an iterable
        my_iterator = iter(my_list)  # my_iterator is an iterator
        
        print(next(my_iterator))  # Output: 1
        print(next(my_iterator))  # Output: 2
        print(next(my_iterator))  # Output: 3
        #print(next(my_iterator))

 6. Explain the concepts of generator in python and how they are defined?
       Ans:-
       In Python, a generator is a special kind of function that returns an iterator, generating values on demand rather than storing them all in memory
        at once. This is achieved using the yield keyword instead of return. Generators are particularly useful for working with large or infinite
        datasets because they produce values one at a time. 
        Definition:
        A generator function in Python is defined like a regular function but utilizes the yield keyword to produce a sequence of values. When called,
        the generator function returns a generator object, which is an iterator. Iterating over the generator object retrieves the values yielded
        by the function, one at a time. 
        Key aspects of generators:
        yield keyword:
        The yield keyword is used to produce a value and pause the generator's execution, preserving its state. 
        Lazy evaluation:
        Generators generate values on demand, only when they are needed, which can be more memory-efficient. 
        Iteration:
        Generators are iterable, meaning they can be traversed using a for loop or the next() function. 
        State preservation:
        The generator maintains its state between calls to next() (or iterations in a loop), allowing it to resume execution where it left off. 
        E.g.
        def count_up_to(n):
            i = 0
            while i < n:
                yield i
                i += 1
        
        # Create a generator object
        counter = count_up_to(5)

        # Iterate over the generator
        for number in counter:
            print(number)

 7. What are the advantages of using generators over regular functions?
   Ans:- 
           Generators in Python offer several advantages over regular functions, primarily in terms of memory efficiency, lazy evaluation, and code
           readability.
    Memory Efficiency:
    Generators produce values on demand using the yield keyword, meaning they don't store the entire sequence in memory. Regular functions,
    on the other hand, typically compute and return an entire collection of results at once, which can be memory-intensive, especially when dealing with 
    large datasets.
    Lazy Evaluation:
    Generators compute values only when they are needed. This lazy evaluation approach can lead to significant performance improvements, especially when
    working with potentially infinite sequences or when only a subset of the results is required. Regular functions evaluate all expressions immediately,
    regardless of whether their results are used.
    Improved Readability:
    Generators can simplify code, especially when dealing with complex iterations or data processing pipelines. By breaking down the logic into smaller,
    more manageable chunks, generators can make the code easier to understand and maintain.
    State Preservation:
    Generators maintain their internal state between calls. This means that each time a generator is called, it resumes execution from where it left off,
    allowing it to produce the next value in the sequence. Regular functions, on the other hand, start execution from the beginning each time they are
    called.
    Suitable for Infinite Sequences:
    Generators can represent infinite sequences, as they only generate values as needed. Regular functions cannot handle infinite sequences directly, 
    as they would need to store all values in memory.

        E.g.
        def regular_function(n):
            result = []
            for i in range(n):
                result.append(i * 2)
            return result
        
        def generator_function(n):
            for i in range(n):
                yield i * 2
        
        #Example usage
        regular_result = regular_function(5)
        print(regular_result) # Output: [0, 2, 4, 6, 8]
        
        generator_result = generator_function(5)
        print(list(generator_result)) # Output: [0, 2, 4, 6, 8]
        
        generator_result = generator_function(5)
        print(next(generator_result)) # Output: 0
        print(next(generator_result)) # Output: 2
        print(next(generator_result)) # Output: 4

  8. What is a lambda function in python and when is it typically used ?
   Ans:-
        In Python, a lambda function is a small, anonymous function that can take any number of arguments but can only have one expression.
        It's essentially a shorthand way to define a function that you'll likely only use once. 
        Lambda functions are typically used in situations where you need a quick, simple function for a specific task, especially when passing 
        functions as arguments to other functions like map(), filter(), or sorted(). They help to keep the code concise and avoid the need for a 
        full function definition when a simple operation is sufficient. 
        
        E.g.
        # Lambda function to square a number
        square = lambda x: x**2
        
        # Using the lambda function with map()
        numbers = [1, 2, 3, 4]
        squared_numbers = list(map(square, numbers))
        print(squared_numbers)  # Output: [1, 4, 9, 16]

  9. Explain the purpose and usage of 'map( )' function in python?
   Ans:- 
       The map() function in Python applies a given function to each item of an iterable (like a list or tuple) and returns an iterator that yields the
        results. This provides a concise way to perform an operation on all elements of an iterable without using explicit loops. 
        Usage:
        Applying a function to each element.
        E.g.
        def square(x):
            return x * x
        
        numbers = [1, 2, 3, 4, 5]
        squared_numbers = map(square, numbers)
        print(list(squared_numbers)) # Output: [1, 4, 9, 16, 25]

  10. What is the difference between map function reduce function and filter functions in python? 
    Ans:- 
        The map(), filter(), and reduce() functions are built-in higher-order functions in Python that operate on iterables (like lists, tuples, etc.).
        They each serve different purposes:
        map():
        Applies a given function to each item in an iterable and returns a new iterable (a map object, which can be converted to a list or tuple) 
        containing the results. It transforms each element individually.
        filter():
        Creates a new iterable containing only the items from the original iterable for which a given function returns True. It selects elements based 
        on a condition.
        reduce():
        Applies a function cumulatively to the items of an iterable, from left to right, to reduce the iterable to a single value. 
        It combines elements sequentially.
        E.g.
        from functools import reduce

        # Example list
        numbers = [1, 2, 3, 4, 5]
        
        # Map: Apply a function to each item in an iterable
        squared_numbers = list(map(lambda x: x**2, numbers))
        print(f"Map Example: Squaring numbers: {squared_numbers}")
        
        # Filter: Create a list of elements for which a function returns true
        even_numbers = list(filter(lambda x: x % 2 == 0, numbers))
        print(f"Filter Example: Filtering even numbers: {even_numbers}")
        
        # Reduce: Apply a rolling computation to sequential pairs of values in a list
        sum_of_numbers = reduce(lambda x, y: x + y, numbers)
        print(f"Reduce Example: Summing numbers: {sum_of_numbers}")
  11. Using pen and paper write the internal mechanism for sum operation using reduce function on this given list [47,11,42, 13] ;
    from functools import reduce

     Ans:-
        # Function to add two numbers
        def sum(x, y):
            return x + y
        
        a = [47, 11, 42, 13]
        res = reduce(sum, a)
        
        print(res)  # Output: 15

'''
```

```python
# PRACTICAL QUESTIONS :: #
# 1. Write a python function that takes a list of numbers as input and return the sum of all even numbers in the list. #
def sum_even_numbers(numbers):

    even_sum = 0
    for num in numbers:
        if num % 2 == 0:
            even_sum += num
    return even_sum

# Example usage:
my_list = [1, 2, 3, 4, 5, 6]
result = sum_even_numbers(my_list)
print(f"The sum of even numbers is: {result}")
```

```python
# 5. Create an iterator class in python that generates the Fibonacci sequence up to a specified number of a terms. #
class FibonacciIterator:
    def __init__(self, limit):
        self.limit = limit
        self.current = 0
        self.next = 1
        self.count = 0

    def __iter__(self):
        return self

    def __next__(self):
        if self.count < self.limit:
            result = self.current
            self.current, self.next = self.next, self.current + self.next
            self.count += 1
            return result
        else:
            raise StopIteration

# Example usage:
n=int(input("Enter how many terms ? "))
fib_iter = FibonacciIterator(n)
for num in fib_iter:
    print(num)
```

```python
# 6. Write a generator function in python that yields the power of two up to a given exponent #
def power_of_two_generator(exponent):
    for i in range(exponent + 1):
        yield 2 ** i

# Example usage
for value in power_of_two_generator(5):
    print(value)
# Expected output: 1, 2, 4, 8, 16, 32
```

```python
# 7. Implement a generator function that reads a file line by line and yields each line as a string #
def read_file_lines(file_path):
    try:
        with open(file_path, 'r') as file:
            for line in file:
                yield line
    except FileNotFoundError:
        print(f"Error: File not found at {file_path}")

# Example usage:
if __name__ == "__main__":
    # Create a dummy text file for testing
    with open("sample.txt", "w") as f:
        f.write("This is the first line.\n")
        f.write("This is the second line.\n")
        f.write("This is the third line.\n")

    # Read the file line by line using the generator
    for line in read_file_lines("sample.txt"):
        print(line, end="")
```

```python
# 9. Write a python program that uses map function to convert a list of temperature from Celsius to Fahrenheit.#
def celsius_to_fahrenheit(celsius):
    return (celsius * 9/5) + 32

celsius_temperatures = [0, 10, 25, 37.5]

fahrenheit_temperatures = list(map(celsius_to_fahrenheit, celsius_temperatures))

print("Celsius Temperatures:", celsius_temperatures)
print("Fahrenheit Temperatures:", fahrenheit_temperatures)
```
